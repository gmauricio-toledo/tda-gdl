#  Transformaciones de preprocesamiento

Generamos puntos aleatorios en el plano para visualizar el efecto de las diferentes transformaciones

## Escalamiento min-max

```python
import numpy as np

# Generamos las coordenadas por separado:
puntos_x = np.random.uniform(low=-3, high=1, size=(30, ))
puntos_y = np.random.uniform(low=-1, high=5, size=(30, ))

# Las unimos:
puntos = np.vstack((puntos_x, puntos_y)).T

print(puntos.shape)
print(puntos)
```

Aplicamos la transformación `min-max`. Usaremos la implementación de scikit-learn [`MinMaxScaler`](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.MinMaxScaler.html).

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
scaler.fit(puntos)
puntos_transformados = scaler.transform(puntos)
print(puntos_transformados)
```

Al ser una clase podemos consultar los atributos de la instancia

```python
print(f"Máximos: {scaler.data_max_}")
print(f"Mínimos: {scaler.data_min_}")
print(f"Rangos: {scaler.data_range_}")
```

```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(1, 2, figsize=(10, 5))
axs[0].scatter(puntos[:,0], puntos[:,1])
axs[0].set_title('Puntos originales')
axs[0].axhline(0, color='black')
axs[0].axvline(0, color='black')
axs[0].set_aspect('equal')
axs[1].scatter(puntos_transformados[:,0], puntos_transformados[:,1])
axs[1].set_title('Puntos transformados')
axs[1].axhline(0, color='black')
axs[1].axvline(0, color='black')
axs[1].set_aspect('equal')
plt.show()
```

## Escalamiento z-score

```python
import numpy as np

# Generamos las variables por separado:
puntos_x = np.random.normal(loc=-3,
                          scale=5,
                          size=(30,))
puntos_y = np.random.normal(loc=1,
                          scale=2,
                          size=(30,))

# Las unimos
puntos = np.vstack((puntos_x, puntos_y)).T

print(puntos)
```

Aplicamos la transformación z-score `StandardScaler`. Usaremos la implementación de scikit-learn [`StandardScaler`](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.StandardScaler.html).

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(puntos)
puntos_transformados = scaler.transform(puntos)
print(puntos_transformados)
```

```python
print(f"Media: {scaler.mean_}")
print(f"Desviación estándar: {scaler.scale_}")
```

```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(1, 2, figsize=(10, 5))
axs[0].scatter(puntos[:,0], puntos[:,1])
axs[0].set_title('Puntos originales')
axs[0].axhline(0, color='black')
axs[0].axvline(0, color='black')
axs[0].set_aspect('equal')
axs[1].scatter(puntos_transformados[:,0], puntos_transformados[:,1])
axs[1].set_title('Puntos transformados')
axs[1].axhline(0, color='black')
axs[1].axvline(0, color='black')
axs[1].set_aspect('equal')
fig.tight_layout()
plt.show()
```

## Normalización unitaria

```python
import numpy as np

puntos = np.random.uniform(low=-3, high=3, size=(30, 2))
print(puntos.shape)
```

Observar que en este caso es una función y no una clase

```python
from sklearn.preprocessing import normalize

puntos_transformados_l2 = normalize(puntos,norm='l2')  # Usando la métrica Euclidiana
puntos_transformados_l1 = normalize(puntos,norm='l1')  # Usando la métrica Manhattan
```

```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(1, 3, figsize=(12,4))
axs[0].scatter(puntos[:,0], puntos[:,1])
axs[0].set_title('Puntos originales')
axs[0].axhline(0, color='black')
axs[0].axvline(0, color='black')
axs[0].set_aspect('equal')
axs[1].scatter(puntos_transformados_l1[:,0], puntos_transformados_l1[:,1])
axs[1].set_title('Puntos transformados L1')
axs[1].axhline(0, color='black')
axs[1].axvline(0, color='black')
axs[1].set_aspect('equal')
axs[2].scatter(puntos_transformados_l2[:,0], puntos_transformados_l2[:,1])
axs[2].set_title('Puntos transformados L2')
axs[2].axhline(0, color='black')
axs[2].axvline(0, color='black')
axs[2].set_aspect('equal')
fig.tight_layout()
plt.show()
```

# Efecto del preprocesamiento en el aprendizaje automático

Generamos el dataset con 2 features, usamos las función [`make_classification`](https://scikit-learn.org/stable/modules/generated/sklearn.datasets.make_classification.html)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=500,          # Número total de muestras/ejemplos a generar
    n_features=2,           # Número total de características/columnas
    n_redundant=0,          # Características redundantes (combinaciones lineales de las informativas)
    n_informative=2,        # Características que realmente contribuyen a la clasificación
    n_clusters_per_class=1, # Número de grupos/clusters por cada clase
    random_state=42         # Semilla para reproducibilidad de los resultados
)

X.shape, y.shape
```

Graficamos

```python
plt.figure()
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis')
plt.axis('equal')
plt.show()
```

Modificamos las escalas para hacer una feature mucho más grande que la otra

```python
X[:, 0] = X[:, 0] * 100  # Feature 1: escala grande (cientos)
X[:, 1] = X[:, 1] * 0.1  # Feature 2: escala muy pequeña (decimas)
```

Graficamos

```python
plt.figure()
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis')
plt.axis('equal')
plt.show()
```

Dividimos los datos

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y,
                                                    test_size=0.25,
                                                    random_state=42)
```

Generamos una versión re-escalada de los datos

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)  # Entrenamos con el conjunto de entrenamiento y lo transformamos
X_test_scaled = scaler.transform(X_test)        # Transformamos el conjunto de prueba (no entrenamos)
```

Graficamos ambas versiones del conjunto de prueba

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Gráfico 1: Datos originales
scatter1 = ax1.scatter(X_train[:, 0], X_train[:, 1], c=y_train, cmap='viridis', alpha=0.7)
ax1.scatter([np.min(X_train),np.max(X_train)], [np.min(X_train),np.max(X_train)], alpha=0.001)
ax1.set_title('Datos Originales')
ax1.grid(True, alpha=0.3)
ax1.set_aspect('equal')
ax1.grid(False)

# Gráfico 2: Datos escalados
scatter2 = ax2.scatter(X_train_scaled[:, 0], X_train_scaled[:, 1], c=y_train, cmap='viridis', alpha=0.7)
ax2.set_title('Datos CON MinMax Scaling')
ax2.grid(True, alpha=0.3)
ax2.set_aspect('equal')
ax2.grid(False)

fig.show()
```

![Pipeline del Machine Learning](https://drive.google.com/uc?id=1S9KVyZbkiciIEeC7cLi-epa62gfFM0pi)

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Entrenar KNN sin scaling
knn_original = KNeighborsClassifier(n_neighbors=5)
knn_original.fit(X_train, y_train)
y_pred_original = knn_original.predict(X_test)

# Entrenar KNN con scaling
knn_scaled = KNeighborsClassifier(n_neighbors=5)
knn_scaled.fit(X_train_scaled, y_train)
y_pred_scaled = knn_scaled.predict(X_test_scaled)

# Calcular métricas de rendimiento
acc_original = accuracy_score(y_test, y_pred_original)
acc_scaled = accuracy_score(y_test, y_pred_scaled)

# Comparar resultados:
print(f"SIN SCALING:")
print(f"Accuracy:  {acc_original:.3f}")

print(f"CON SCALING:")
print(f"Accuracy:  {acc_scaled:.3f}")

print(f"\nMEJORAS:")
print(f"Diferencia en Accuracy:  {acc_scaled - acc_original}")
print(f"Mejora proporcional: {(acc_scaled - acc_original)/acc_original}")
```

Graficamos ambas métricas

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 6))

bars = ax.bar([1, 2], [acc_original, acc_scaled], color=['skyblue', 'orange'])

ax.set_xticks([1, 2])
ax.set_xticklabels(['Original', 'Escalado'])
plt.ylabel('Accuracy')
plt.suptitle('Comparación de Métricas')
plt.grid(False)
plt.show()
```
