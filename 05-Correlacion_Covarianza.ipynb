# Correlaciones y covarianza

Calculemos la covarianza de varios conjuntos de datos pequeños

```python
import numpy as np
import matplotlib.pyplot as plt

xs = np.array([1, 2, 3, 4, 5,6], dtype=float)
# xs = -np.array([1, 2, 3, 4, 5, 6], dtype=float)
ys = 2*xs  # Relación directamente proporcional entre ambas variables
# ys = 2*xs + np.random.normal(size=xs.shape) # Con ruido
# ys = -2*xs # Relación inversamente proporcional
# ys = -2*xs + np.random.normal(size=xs.shape,scale=1)
ys = xs**2 + 1


print(xs)
print(ys)
print(f"Matriz de covarianza:\n{np.cov(xs,ys)}")

plt.figure()
plt.scatter(xs, ys)
plt.show()

```

```python
puntos = np.random.uniform(size=(100,2))

plt.figure()
plt.scatter(puntos[:,0], puntos[:,1])
plt.show()

print(f"Matriz de covarianza:\n{np.cov(puntos[:,0],puntos[:,1])}")
```

```python
xs = np.random.uniform(size=100)
ys = xs**2 + 1 + np.random.normal(size=xs.shape,scale=0.25)

print(f"Matriz de covarianza:\n{np.cov(xs,ys)}")

plt.figure()
plt.scatter(xs, ys)
plt.show()
```

## Correlación

```python
import pandas as pd

housing_prices_df = pd.read_csv('/content/sample_data/california_housing_train.csv')
housing_prices_df
```

```python
housing_prices_df.corr()
```

```python
import matplotlib.pyplot as plt
from seaborn import heatmap

corrs = housing_prices_df.corr().values

plt.figure()
heatmap(corrs, annot=True,cmap='viridis')
plt.xticks(np.array(range(len(housing_prices_df.columns)))+0.5, housing_prices_df.columns,
           rotation=90)
plt.yticks(np.array(range(len(housing_prices_df.columns)))+0.5, housing_prices_df.columns,
           rotation=0)
plt.show()
```
