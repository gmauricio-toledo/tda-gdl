# Correlaciones y covarianza

Calculemos la covarianza de varios conjuntos de datos pequeños

```python
import numpy as np
import matplotlib.pyplot as plt

xs = np.array([1, 2, 3, 4, 5,6], dtype=float)
# xs = -np.array([1, 2, 3, 4, 5, 6], dtype=float)
ys = 2*xs  # Relación directamente proporcional entre ambas variables
# ys = 2*xs + np.random.normal(size=xs.shape) # Con ruido
# ys = -2*xs # Relación inversamente proporcional
# ys = -2*xs + np.random.normal(size=xs.shape,scale=1)
ys = xs**2 + 1


print(xs)
print(ys)
print(f"Matriz de covarianza:\n{np.cov(xs,ys)}")

plt.figure()
plt.scatter(xs, ys)
plt.show()

```

```python
puntos = np.random.uniform(size=(500,2),low=0,high=10)

plt.figure()
plt.scatter(puntos[:,0], puntos[:,1])
plt.show()

print(f"Matriz de covarianza:\n{np.cov(puntos[:,0],puntos[:,1])}")
```

```python
from re import M
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs, make_moons

xs, ys = [], []
size = 500

#------------ Creamos los datasets ---------------
xs_1 = np.random.normal(size=size)
ys_1 = 0.5*xs_1 - 1
xs.append(xs_1)
ys.append(ys_1)

xs_2 = np.random.normal(size=size)
ys_2 = 0.5*xs_2 + 1 + np.random.normal(size=size,scale=0.3)
xs.append(xs_2)
ys.append(ys_2)

xs_3 = np.random.normal(size=size)
ys_3 = 0.5*xs_3 + 1 + np.random.normal(size=size,scale=1.5)
xs.append(xs_3)
ys.append(ys_3)

xs_4 = np.random.normal(size=size)
ys_4 = np.random.normal(size=size)
xs.append(xs_4)
ys.append(ys_4)

xs_5 = xs_1
ys_5 = -0.5*xs_1 + 1
xs.append(xs_5)
ys.append(ys_5)

xs_6 = xs_1
ys_6 = np.zeros_like(xs_1)
xs.append(xs_6)
ys.append(ys_6)

xs_7 = np.random.uniform(size=size,low=-1,high=1)
ys_7 = np.random.uniform(size=size,low=-1,high=1)
xs.append(xs_7)
ys.append(ys_7)

blobs = make_blobs(n_samples=size,centers=3)
xs_8 = blobs[0][:,0]
ys_8 = blobs[0][:,1]
xs.append(xs_8)
ys.append(ys_8)

moons, labels = make_moons(n_samples=size,noise=0.1)
xs_9 = moons[labels==0,0]
ys_9 = moons[labels==0,1]
xs.append(xs_9)
ys.append(ys_9)

num_datasets = len(xs)
cols = 3
if num_datasets%cols == 0:
    rows = num_datasets//cols
else:
    rows = num_datasets//cols + 1

fig, axs = plt.subplots(rows, cols, figsize=(5*cols, 5*rows))
for j,ax in enumerate(axs.flatten()):
    if j < num_datasets:
        cov = np.cov(xs[j],ys[j])[0,1]
        ax.scatter(xs[j], ys[j],color='red')
        ax.set_title(round(cov,2),fontsize=20)
        ax.axhline(0,color='k')
        ax.axvline(0,color='k')
        ax.axis('off')
fig.tight_layout()
fig.show()
```

## Correlación

```python
import pandas as pd

housing_prices_df = pd.read_csv('/content/sample_data/california_housing_train.csv')
housing_prices_df
```

```python
housing_prices_df.corr()
```

```python
import matplotlib.pyplot as plt
from seaborn import heatmap

corrs = housing_prices_df.corr().values

plt.figure()
heatmap(corrs, annot=True,cmap='viridis')
plt.xticks(np.array(range(len(housing_prices_df.columns)))+0.5, housing_prices_df.columns,
           rotation=90)
plt.yticks(np.array(range(len(housing_prices_df.columns)))+0.5, housing_prices_df.columns,
           rotation=0)
plt.show()
```
